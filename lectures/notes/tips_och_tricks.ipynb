Vår uppgift är följande:

Vi ska bestämma procentuell ökning av lön för varje anställd. Ju högre lön du har, desto större procentuell ökning ska du få.

50000 < lön =< 60000 --> 5%

60000 < lön <= 70000 --> 4%

70000 < lön <= 80000 --> 3%

80000 < lön <= 90000 --> 2%


```python
def procentuellt_ökning(lön: int | float) -> str:

    """Den här funktionen tar in en lön, och spottar ur sig procentuell ökning vid nästa lönelyft"""

    if 50000 < lön <= 60000:
        return "5%"
    
    elif 60000 < lön <= 70000:
        return "4%"
    
    elif 70000 < lön <= 80000:
        return "3%"
    
    elif 80000 < lön <= 90000:
        return "2%"
```

```python
df['Procentuell ökning (sträng)'] = df['Salary'].apply(procentuellt_ökning)
```

```python
df
```

```python
def procentuellt_ökning_v2(lön: int | float) -> float:

    """Den här funktionen tar in en lön, och spottar ur sig procentuell ökning vid nästa lönelyft"""

    if 50000 < lön <= 60000:
        return 1.05
    
    elif 60000 < lön <= 70000:
        return 1.04
    
    elif 70000 < lön <= 80000:
        return 1.03
    
    elif 80000 < lön <= 90000:
        return 1.02
```

```python
df['Procentuell ökning'] = df['Salary'].apply(procentuellt_ökning_v2)

df
```

```python
df['Salary after increase'] = df['Salary'] * df['Procentuell ökning']

df
```
